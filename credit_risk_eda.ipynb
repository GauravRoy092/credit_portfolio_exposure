#The Baseline Setup (Defining "Default")

```python
# 1. Filter the dataset to ONLY include loans that have finished (Success or Failure)
closed_statuses = ['Fully Paid', 'Charged Off']
df_closed = df_cleaned[df_cleaned['loan_status'].isin(closed_statuses)].copy()

# 2. Create our binary math column (1 = Default/Charged Off, 0 = Success/Fully Paid)
# We use an easy trick: check if it equals 'Charged Off', then turn that True/False into 1/0
df_closed['default_flag'] = (df_closed['loan_status'] == 'Charged Off').astype(int)

print("Total Closed Loans ready for math:", len(df_closed))
```

```python
#Income vs. Default Risk

import pandas as pd

# 1. We cannot group by exact income (every dollar amount is different). 
# We must create "Income Tiers" or "Bins".
bins = [0, 50000, 100000, 150000, float('inf')]
labels = ['1. Low (<$50k)', '2. Medium ($50k-$100k)', '3. High ($100k-$150k)', '4. Very High (>$150k)']

df_closed['income_tier'] = pd.cut(df_closed['annual_inc'], bins=bins, labels=labels)

# 2. Group the data by our new tiers, and calculate the MEAN of the default_flag.
# Because defaults are 1s and successes are 0s, the mean mathematically equals the exact default percentage!
income_risk = df_closed.groupby('income_tier', observed=False)['default_flag'].mean() * 100

# 3. Rename the column so it looks like a professional report
income_risk_table = income_risk.reset_index()
income_risk_table.columns = ['Income Tier', 'Default Rate (%)']

# 4. Show the final calculated table
print(income_risk_table)
```

```python
#Collateral & Housing Stability. How do we check if renters are riskier than homeowners?


import pandas as pd

# 1. Filter out rare or noise categories like 'OTHER', 'NONE', 'ANY'
major_housing = ['MORTGAGE', 'RENT', 'OWN']
df_housing = df_closed[df_closed['home_ownership'].isin(major_housing)]

# 2. Aggregate metrics: Total Loans, Average Loan Amount, and Default Rate (%)
housing_risk = df_housing.groupby('home_ownership', observed=False).agg(
    total_loans=('default_flag', 'count'),
    avg_loan_amnt=('loan_amnt', 'mean'),
    default_rate=('default_flag', lambda x: x.mean() * 100)
).reset_index()

# 3. Format values for a polished executive presentation
housing_risk['default_rate'] = housing_risk['default_rate'].round(2)
housing_risk['avg_loan_amnt'] = housing_risk['avg_loan_amnt'].round(2)
housing_risk = housing_risk.sort_values(by='default_rate', ascending=False)

# 4. Display the table
print(housing_risk)
```

```python
import pandas as pd

# 1. Group the data by 'purpose' and calculate total volume, average loan size, and default rate
purpose_risk = df_closed.groupby('purpose', observed=False).agg(
    total_loans=('default_flag', 'count'),
    avg_loan_amnt=('loan_amnt', 'mean'),
    default_rate=('default_flag', lambda x: x.mean() * 100)
).reset_index()

# 2. Format the numbers for a clean read
purpose_risk['default_rate'] = purpose_risk['default_rate'].round(2)
purpose_risk['avg_loan_amnt'] = purpose_risk['avg_loan_amnt'].round(2)

# 3. Sort from the most dangerous (highest default rate) to the safest
purpose_risk = purpose_risk.sort_values(by='default_rate', ascending=False)

# 4. Display the results
print(purpose_risk)
```

```python
# If Plotly is not pre-installed in your Anaconda, you can install it by running this in a cell:
# !pip install plotly

import plotly.express as px

# 1. We feed Plotly Express (px) the purpose_risk table we made in Phase 3
fig = px.bar(
    purpose_risk, 
    x='default_rate', 
    y='purpose', 
    orientation='h',  # Horizontal bar chart
    title='Interactive Risk Dashboard: Default Rate by Loan Purpose',
    color='default_rate',           # Color the bars based on how dangerous they are
    color_continuous_scale='Reds',  # Use a heatmap gradient (Dark Red = High Risk)
    text='default_rate',            # Print the exact percentage on the bar itself
    
    # This is the magic of Plotly: When a recruiter hovers their mouse, show these extra details!
    hover_data={'total_loans': True, 'avg_loan_amnt': True, 'default_rate': False}
)

# 2. Clean up the visual layout and sort the bars from highest to lowest risk
fig.update_layout(
    yaxis={'categoryorder': 'total ascending'}, 
    xaxis_title='Default Rate (%)',
    yaxis_title='Loan Purpose',
    height=700 # Give it enough height so all categories fit perfectly
)

# 3. Render the interactive dashboard component
fig.show()
```

```python
# If ipywidgets is not installed, you can run: !pip install ipywidgets
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display
import pandas as pd

# 1. Ensure we have the 'issue_year' column in our closed loans dataset
if 'issue_year' not in df_closed.columns:
    df_closed['issue_year'] = pd.to_datetime(df_closed['issue_d']).dt.year

# 2. Get a list of all the years in the dataset, and add an "All Years" option
available_years = sorted(df_closed['issue_year'].dropna().unique())
dropdown_options = ['All Years'] + list(available_years)

# 3. Create the function that updates everything when a new year is selected
def update_dashboard(selected_year):
    
    # Step A: Filter the data based on the dropdown choice
    if selected_year == 'All Years':
        filtered_df = df_closed
    else:
        filtered_df = df_closed[df_closed['issue_year'] == selected_year]
        
    # Step B: Recalculate the risk metrics for just that specific year
    purpose_risk = filtered_df.groupby('purpose', observed=False).agg(
        total_loans=('default_flag', 'count'),
        avg_loan_amnt=('loan_amnt', 'mean'),
        default_rate=('default_flag', lambda x: x.mean() * 100)
    ).reset_index()
    
    purpose_risk['default_rate'] = purpose_risk['default_rate'].round(2)
    purpose_risk = purpose_risk.sort_values(by='default_rate', ascending=False)
    
    # Step C: Draw the Plotly chart with the new filtered data
    fig = px.bar(
        purpose_risk, 
        x='default_rate', 
        y='purpose', 
        orientation='h',
        title=f'Risk by Loan Purpose ({selected_year})',
        color='default_rate',
        color_continuous_scale='Reds',
        text='default_rate'
    )
    
    fig.update_layout(
        yaxis={'categoryorder': 'total ascending'}, 
        xaxis_title='Default Rate (%)',
        yaxis_title='Loan Purpose',
        height=600
    )
    
    fig.show()

# 4. Bind the dropdown menu to our update function
widgets.interact(update_dashboard, selected_year=dropdown_options);
```
